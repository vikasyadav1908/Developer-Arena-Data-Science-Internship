```python
# Student Grade Calculator

def calculate_grade(marks):
    """
    Calculate grade based on student's marks.
    """

    if marks >= 90:
        grade = "A"
        message = "Excellent! Outstanding performance! "

    elif marks >= 80:
        grade = "B"
        message = "Very Good! Keep it up! "

    elif marks >= 70:
        grade = "C"
        message = "Good job! You can improve further! "

    elif marks >= 60:
        grade = "D"
        message = "You passed! Keep working harder! "

    else:
        grade = "F"
        message = "Don't give up! Work harder and try again! "

    return grade, message


def get_valid_marks():
    """
    Get valid marks between 0 and 100.
    """

    while True:
        try:
            marks = float(input("Enter marks (0-100): "))

            if 0 <= marks <= 100:
                return marks

            else:
                print("❌ Invalid marks!")
                print("Please enter marks between 0 and 100.\n")

        except ValueError:
            print("❌ Invalid input!")
            print("Please enter numbers only.\n")


def main():
    # Get student name
    student_name = input("Enter student name: ")

    # Get valid marks
    marks = get_valid_marks()

    # Calculate grade
    grade, message = calculate_grade(marks)

    # Display result
    print("\n" + "=" * 35)
    print(f"📊 RESULT FOR {student_name.upper()}")
    print("=" * 35)

    print(f"Marks: {marks}/100")
    print(f"Grade: {grade}")
    print(f"Message: {message}")

    print("=" * 35)


# Start the program
if __name__ == "__main__":
    main()
```
