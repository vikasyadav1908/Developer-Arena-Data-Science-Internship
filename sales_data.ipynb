```python
import pandas as pd

# Load Dataset
df = pd.read_csv("sales_data.csv")
```

```python
# Explore Dataset
print("First 5 Rows")
print(df.head())
```

```python

print("Shape")
print(df.shape)
```

```python
print("Columns")
print(df.columns)

print("Data Types")
print(df.dtypes)

print("Missing Values")
print(df.isnull().sum())

print("Duplicate Rows")
print(df.duplicated().sum())

```

```python
# Data Cleaning
df = df.drop_duplicates()

df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Price"] = df["Price"].fillna(df["Price"].median())
df["Total_Sales"] = df["Total_Sales"].fillna(df["Quantity"] * df["Price"])
```

```python

# Analysis
total_revenue = df["Total_Sales"].sum()
average_sale = df["Total_Sales"].mean()
total_quantity = df["Quantity"].sum()

best_product = df.groupby("Product")["Quantity"].sum()
highest_revenue = df.groupby("Product")["Total_Sales"].sum()

product_sales = df.groupby("Product")["Total_Sales"].sum()
region_sales = df.groupby("Region")["Total_Sales"].sum()
```

```python
# Final Report
print("==================================================")
print("FINAL SALES REPORT")
print("==================================================")

print("Total Revenue:", total_revenue)
print("Average Sale:", average_sale)
print("Total Quantity Sold:", total_quantity)
print("Best Selling Product:", best_product.idxmax())
print("Highest Revenue Product:", highest_revenue.idxmax())

print("Revenue by Product")
print(product_sales)

print("Sales by Region")
print(region_sales)

print("Analysis Completed Successfully")
```
